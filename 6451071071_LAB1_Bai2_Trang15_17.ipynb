```python
# In họ tên và MSSV
print("Họ và tên: Trần Minh Tấn")
print("MSSV: 6451071071")

import pandas as pd
import io
import requests

# Tải dữ liệu từ Google Drive
file_id = '1TCnJqaowkv0caJTJrmjoqb0AB2bMvgxP'
download_url = f'https://drive.google.com/uc?export=download&id={file_id}'
response = requests.get(download_url)
df = pd.read_csv(io.StringIO(response.content.decode('utf-8')))

print("--- KÍCH THƯỚC BAN ĐẦU CỦA DỮ LIỆU ---")
print(df.shape)

print("\n--- SỐ LƯỢNG GIÁ TRỊ THIẾU BAN ĐẦU ---")
print(df.isnull().sum())

# 1. Xóa bỏ các thuộc tính không cần thiết hoặc chứa quá nhiều giá trị rỗng/không phục vụ cho phân tích (ví dụ: 'cabin', 'boat', 'body', 'home.dest', 'name', 'ticket')
columns_to_drop = ['cabin', 'boat', 'body', 'home.dest', 'name', 'ticket']
df = df.drop(columns=[col for col in columns_to_drop if col in df.columns])

print("\n--- CÁC CỘT CÒN LẠI SAU KHI XÓA ---")
print(df.columns.tolist())

# 2. Xử lý các giá trị thiếu (Missing values) cho các thuộc tính quan trọng còn lại
# - Điền giá trị thiếu của thuộc tính 'age' bằng giá trị trung bình (mean)
if 'age' in df.columns:
    df['age'] = df['age'].fillna(df['age'].mean())

# - Điền giá trị thiếu của thuộc tính 'fare' bằng giá trị trung bình (mean)
if 'fare' in df.columns:
    df['fare'] = df['fare'].fillna(df['fare'].mean())

# - Điền giá trị thiếu của thuộc tính 'embarked' bằng giá trị xuất hiện nhiều nhất (mode)
if 'embarked' in df.columns:
    df['embarked'] = df['embarked'].fillna(df['embarked'].mode()[0])

print("\n--- SỐ LƯỢNG GIÁ TRỊ THIẾU SAU KHI TIỀN XỬ LÝ ---")
print(df.isnull().sum())

print("\n--- KÍCH THƯỚC DỮ LIỆU SAU KHI TIỀN XỬ LÝ HOÀN TẤT ---")
print(df.shape)
display(df.head())
```

```python
# In họ tên và MSSV
print("Họ và tên: Trần Minh Tấn")
print("MSSV: 6451071071")

import pandas as pd
import numpy as np
import io
import requests

# Tải dữ liệu từ Google Drive
file_id = '1TCnJqaowkv0caJTJrmjoqb0AB2bMvgxP'
download_url = f'https://drive.google.com/uc?export=download&id={file_id}'
response = requests.get(download_url)
df = pd.read_csv(io.StringIO(response.content.decode('utf-8')))

print("--- KÍCH THƯỚC BAN ĐẦU ---")
print(df.shape)

# 1. Điền giá trị thiếu cơ bản trước khi biến đổi (tránh lỗi NaN khi ép kiểu/rời rạc hóa)
if 'age' in df.columns:
    df['age'] = df['age'].fillna(df['age'].mean())
if 'fare' in df.columns:
    df['fare'] = df['fare'].fillna(df['fare'].mean())
if 'sex' in df.columns:
    df['sex'] = df['sex'].fillna(df['sex'].mode()[0])
if 'embarked' in df.columns:
    df['embarked'] = df['embarked'].fillna(df['embarked'].mode()[0])

# 2. Trích xuất thuộc tính danh hiệu (title) từ tên (name)
# Ví dụ: 'Braund, Mr. Owen Harris' -> 'Mr'
if 'name' in df.columns:
    df['title'] = df['name'].str.extract(' ([A-Za-z]+)\.', expand=False)
    # Gom nhóm các title hiếm gặp nếu cần, hoặc giữ nguyên
    df['title'] = df['title'].replace(['Lady', 'Countess','Capt', 'Col',\
                                       'Don', 'Dr', 'Major', 'Rev', 'Sir', 'Jonkheer', 'Dona'], 'Rare')
    df['title'] = df['title'].replace('Mlle', 'Miss')
    df['title'] = df['title'].replace('Ms', 'Miss')
    df['title'] = df['title'].replace('Mme', 'Mrs')

# 3. Chuyển đổi thuộc tính giới tính (sex) và cổng lên tàu (embarked) thành số
df['sex'] = df['sex'].map({'female': 0, 'male': 1})
df['embarked'] = df['embarked'].map({'C': 0, 'Q': 1, 'S': 2})

# 4. Chia giá trị thuộc tính tuổi (age) thành 8 nhóm theo gợi ý
# Nhóm 1: <= 11 | Nhóm 2: (11, 18] | Nhóm 3: (18, 22] | Nhóm 4: (22, 27]
# Nhóm 5: (27, 33] | Nhóm 6: (33, 40] | Nhóm 7: (40, 66] | Nhóm 8: > 66
age_bins = [-1, 11, 18, 22, 27, 33, 40, 66, 120]
age_labels = [1, 2, 3, 4, 5, 6, 7, 8]
df['age_group'] = pd.cut(df['age'], bins=age_bins, labels=age_labels)

# 5. Chia giá trị thuộc tính giá vé (fare) thành 6 nhóm theo gợi ý
# Nhóm 1: <= 7.91 | Nhóm 2: (7.91, 14.454] | Nhóm 3: (14.454, 31]
# Nhóm 4: (31, 99] | Nhóm 5: (99, 250] | Nhóm 6: > 250
fare_bins = [-float('inf'), 7.91, 14.454, 31, 99, 250, float('inf')]
fare_labels = [1, 2, 3, 4, 5, 6]
df['fare_group'] = pd.cut(df['fare'], bins=fare_bins, labels=fare_labels)

# 6. Chuyển đổi thuộc tính giá vé (fare) về kiểu nguyên (sau khi đã xử lý hoặc chuyển đổi qua nhóm)
df['fare'] = df['fare'].astype(int)

print("\n--- 5 DÒNG DỮ LIỆU SAU KHI BIẾN ĐỔI ---")
display(df[['name', 'title', 'sex', 'embarked', 'age', 'age_group', 'fare', 'fare_group']].head())

print("\n--- THÔNG TIN KIỂU DỮ LIỆU CÁC CỘT MỚI ---")
print(df[['title', 'sex', 'embarked', 'age_group', 'fare', 'fare_group']].dtypes)
```

```python
# In họ tên và MSSV
print("Họ và tên: Trần Minh Tấn")
print("MSSV: 6451071071")

import pandas as pd
import numpy as np
import io
import requests

# Tải dữ liệu từ Google Drive
file_id = '1TCnJqaowkv0caJTJrmjoqb0AB2bMvgxP'
download_url = f'https://drive.google.com/uc?export=download&id={file_id}'
response = requests.get(download_url)
df = pd.read_csv(io.StringIO(response.content.decode('utf-8')))

# Xử lý giá trị thiếu cơ bản cho các cột liên quan
if 'age' in df.columns:
    df['age'] = df['age'].fillna(df['age'].mean())
if 'fare' in df.columns:
    df['fare'] = df['fare'].fillna(df['fare'].mean())

# 1. Tạo thuộc tính tuổi * loại vé (age_class): là tích của tuổi (age) và loại vé (pclass)[cite: 14]
if 'age' in df.columns and 'pclass' in df.columns:
    df['age_class'] = df['age'] * df['pclass']

# 2. Tạo thuộc tính giá vé trên mỗi người (fare_per_person): là giá vé chia cho số người đi cùng + 1 (có làm tròn)[cite: 14]
# Tổng số người đi cùng (FamilySize) = sibsp + parch
if 'fare' in df.columns and 'sibsp' in df.columns and 'parch' in df.columns:
    df['family_size'] = df['sibsp'] + df['parch'] + 1
    df['fare_per_person'] = df['fare'] / df['family_size']
    # Làm tròn giá vé trên mỗi người (nếu cần thiết theo dạng số nguyên hoặc làm tròn chữ số thập phân)
    df['fare_per_person'] = df['fare_per_person'].round(2)

print("\n--- 5 DÒNG DỮ LIỆU SAU KHI TẠO CÁC THUỘC TÍNH MỚI ---")
display(df[['name', 'age', 'pclass', 'age_class', 'fare', 'sibsp', 'parch', 'fare_per_person']].head())

print("\n--- THÔNG TIN KIỂU DỮ LIỆU CÁC CỘT MỚI ---")
print(df[['age_class', 'fare_per_person']].dtypes)
```
