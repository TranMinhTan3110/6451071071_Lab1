```python
# In họ tên và MSSV
print("Họ và tên: Trần Minh Tấn")
print("MSSV: 6451071071")

import pandas as pd
import io
import requests

try:
    df = pd.read_excel('Online Retail.xlsx')
except Exception:
    file_id = '12q4inbGbTS1UKs9sHA1X87vSRV8feEHV'
    download_url = f'https://drive.google.com/uc?export=download&id={file_id}'
    response = requests.get(download_url)
    df = pd.read_excel(io.BytesIO(response.content))

df['Description'] = df['Description'].astype(str).str.strip()
df = df.dropna(subset=['InvoiceNo'])
df['InvoiceNo'] = df['InvoiceNo'].astype(str)
df = df[~df['InvoiceNo'].str.contains('C')]

# Gom nhóm dữ liệu cho nước Anh
basket = (df[df['Country'] == 'United Kingdom']
          .groupby(['InvoiceNo', 'Description'])['Quantity']
          .sum().unstack().reset_index().fillna(0).set_index('InvoiceNo'))

# 6. Chuyển đổi dữ liệu về dạng Hot Encoding (0: không mua, 1: có mua)
def encode_units(x):
    if x <= 0:
        return 0
    if x >= 1:
        return 1

# Sử dụng hàm map (hoặc applymap) để biến đổi các giá trị số lượng thành 0 và 1
basket_sets = basket.map(encode_units)

print("--- CÂU 6: DỮ LIỆU SAU KHI DÙNG HOT ENCODING ---")
print("Kích thước ma trận Hot Encoding:", basket_sets.shape)
display(basket_sets.head())
```

```python
# In họ tên và MSSV
print("Họ và tên: Trần Minh Tấn")
print("MSSV: 6451071071")

import pandas as pd
import io
import requests

try:
    df = pd.read_excel('Online Retail.xlsx')
except Exception:
    file_id = '12q4inbGbTS1UKs9sHA1X87vSRV8feEHV'
    download_url = f'https://drive.google.com/uc?export=download&id={file_id}'
    response = requests.get(download_url)
    df = pd.read_excel(io.BytesIO(response.content))

df['Description'] = df['Description'].astype(str).str.strip()
df = df.dropna(subset=['InvoiceNo'])
df['InvoiceNo'] = df['InvoiceNo'].astype(str)
df = df[~df['InvoiceNo'].str.contains('C')]

basket = (df[df['Country'] == 'United Kingdom']
          .groupby(['InvoiceNo', 'Description'])['Quantity']
          .sum().unstack().reset_index().fillna(0).set_index('InvoiceNo'))

# 7. Chuyển đổi dữ liệu từ dạng hot encoding thành one-hot encoding (chuẩn hóa giá trị về kiểu boolean/0-1 hoàn toàn)
def encode_units(x):
    return 1 if x >= 1 else 0

basket_sets = basket.map(encode_units)
# Chuyển đổi tường minh sang kiểu boolean hoặc giữ nguyên số nguyên nhị phân (one-hot)
basket_sets = basket_sets.astype(bool)

print("--- CÂU 7: DỮ LIỆU SAU KHI CHUYỂN THÀNH ONE-HOT ENCODING ---")
print(basket_sets.shape)
display(basket_sets.head())
```

```python
# In họ tên và MSSV
print("Họ và tên: Trần Minh Tấn")
print("MSSV: 6451071071")

import pandas as pd
import warnings
from mlxtend.frequent_patterns import apriori
import io
import requests


warnings.filterwarnings('ignore')

try:
    df = pd.read_excel('Online Retail.xlsx')
except Exception:
    file_id = '12q4inbGbTS1UKs9sHA1X87vSRV8feEHV'
    download_url = f'https://drive.google.com/uc?export=download&id={file_id}'
    response = requests.get(download_url)
    df = pd.read_excel(io.BytesIO(response.content))

df['Description'] = df['Description'].astype(str).str.strip()
df = df.dropna(subset=['InvoiceNo'])
df['InvoiceNo'] = df['InvoiceNo'].astype(str)
df = df[~df['InvoiceNo'].str.contains('C')]
df = df[df['Description'] != 'POSTAGE']

basket = (df[df['Country'] == 'United Kingdom']
          .groupby(['InvoiceNo', 'Description'])['Quantity']
          .sum().unstack().reset_index().fillna(0).set_index('InvoiceNo'))

basket_sets = basket.map(lambda x: 1 if x >= 1 else 0).astype(bool)

# 9. Tìm tập phổ biến bằng thuật toán Apriori với min_sup = 0.03 (3%)
frequent_itemsets_apriori = apriori(basket_sets, min_support=0.03, use_colnames=True)

print("--- CÂU 9: TẬP PHỔ BIẾN SỬ DỤNG APRIORI (min_sup = 3%) ---")
print(f"Tổng số tập phổ biến tìm được: {len(frequent_itemsets_apriori)}")
display(frequent_itemsets_apriori.head(10))
```

```python
# In họ tên và MSSV
print("Họ và tên: Trần Minh Tấn")
print("MSSV: 6451071071")

import pandas as pd
import time
from mlxtend.frequent_patterns import fpgrowth, association_rules
import io
import requests

try:
    df = pd.read_excel('Online Retail.xlsx')
except Exception:
    file_id = '12q4inbGbTS1UKs9sHA1X87vSRV8feEHV'
    download_url = f'https://drive.google.com/uc?export=download&id={file_id}'
    response = requests.get(download_url)
    df = pd.read_excel(io.BytesIO(response.content))

df['Description'] = df['Description'].astype(str).str.strip()
df = df.dropna(subset=['InvoiceNo'])
df['InvoiceNo'] = df['InvoiceNo'].astype(str)
df = df[~df['InvoiceNo'].str.contains('C')]
df = df[df['Description'] != 'POSTAGE']

basket = (df[df['Country'] == 'United Kingdom']
          .groupby(['InvoiceNo', 'Description'])['Quantity']
          .sum().unstack().reset_index().fillna(0).set_index('InvoiceNo'))

basket_sets = basket.map(lambda x: 1 if x >= 1 else 0).astype(bool)

# 12. Tìm tập phổ biến và luật kết hợp bằng thuật toán FP-Growth với min_sup = 3%, min_conf = 50%
start_time = time.time()
frequent_itemsets_fp = fpgrowth(basket_sets, min_support=0.03, use_colnames=True)
rules_fp = association_rules(frequent_itemsets_fp, metric="confidence", min_threshold=0.5)
end_time = time.time()

print("--- CÂU 12: KẾT QUẢ THUẬT TOÁN FP-GROWTH ---")
print(f"Số lượng tập phổ biến (FP-Growth): {len(frequent_itemsets_fp)}")
print(f"Số lượng luật kết hợp (FP-Growth): {len(rules_fp)}")
print(f"Thời gian thực thi FP-Growth: {end_time - start_time:.4f} giây")

print("\n--- SO SÁNH VỚI APRIORI ---")
print("- Cả hai thuật toán đều cho ra kết quả tập phổ biến và các luật kết hợp tương đương nhau vì cùng dựa trên một tập dữ liệu và ngưỡng tối thiểu.")
print("- Thuật toán FP-Growth thường có ưu thế về tốc độ thực thi (nhanh hơn Apriori) do không phải sinh tập ứng viên qua nhiều vòng mà nén dữ liệu vào cây FP-Tree.")
```
